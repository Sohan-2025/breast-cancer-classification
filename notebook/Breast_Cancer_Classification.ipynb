## 4. Normality Testing

```python
from scipy.stats import shapiro

results = []

for feature in X_train.columns:

    benign_values = X_train.loc[y_train == 0, feature]
    malignant_values = X_train.loc[y_train == 1, feature]

    benign_stat, benign_p = shapiro(benign_values)
    malignant_stat, malignant_p = shapiro(malignant_values)

    results.append({
        "feature": feature,
        "benign_p": benign_p,
        "malignant_p": malignant_p
    })

normality_results = pd.DataFrame(results)

normality_results
```

```python
normality_results["benign_normal"] = (
    normality_results["benign_p"] > 0.05
)

normality_results["malignant_normal"] = (
    normality_results["malignant_p"] > 0.05
)

normality_results
```

## 5. Welch Two-Sample t-Test

```python
from scipy.stats import ttest_ind

hypothesis_results = []

for feature in X_train.columns:

    benign = X_train.loc[y_train == 0, feature]
    malignant = X_train.loc[y_train == 1, feature]

    statistic, p_value = ttest_ind(
        benign,
        malignant,
        equal_var=False
    )

    hypothesis_results.append({
        "feature": feature,
        "benign_mean": benign.mean(),
        "malignant_mean": malignant.mean(),
        "t_statistic": statistic,
        "p_value": p_value
    })

hypothesis_results = pd.DataFrame(hypothesis_results)

hypothesis_results
```

```python
alpha = 0.05
hypothesis_results["significant"] = hypothesis_results["p_value"] < alpha

print(f"Significant features at α = {alpha}: {hypothesis_results["significant"].sum()} / {len(hypothesis_results)}")
print("\nNon-significant features:")
display(hypothesis_results.loc[~hypothesis_results["significant"], ["feature", "p_value"]].round(6))
```

### Exploratory Threshold Analysis on the Test Set

```python
import pandas as pd
import numpy as np
from sklearn.metrics import confusion_matrix

thresholds = [0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]

threshold_results = []

for threshold in thresholds:

    # Convert probabilities into class predictions
    y_pred_threshold = (y_prob_lr[:, 1] >= threshold).astype(int)

    # Confusion matrix
    cm = confusion_matrix(y_test, y_pred_threshold)

    tn, fp, fn, tp = cm.ravel()

    # Metrics
    sensitivity = tp / (tp + fn)
    specificity = tn / (tn + fp)
    accuracy = (tp + tn) / (tp + tn + fp + fn)

    threshold_results.append({
        "threshold": threshold,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,
        "accuracy": accuracy,
        "sensitivity": sensitivity,
        "specificity": specificity
    })

threshold_results = pd.DataFrame(threshold_results)

threshold_results
```

## 9. Multicollinearity and Feature Reduction

```python
feature_names = [
    "radius_mean", "texture_mean", "perimeter_mean", "area_mean",
    "smoothness_mean", "compactness_mean", "concavity_mean",
    "concave_points_mean", "symmetry_mean", "fractal_dimension_mean",

    "radius_se", "texture_se", "perimeter_se", "area_se",
    "smoothness_se", "compactness_se", "concavity_se",
    "concave_points_se", "symmetry_se", "fractal_dimension_se",

    "radius_worst", "texture_worst", "perimeter_worst", "area_worst",
    "smoothness_worst", "compactness_worst", "concavity_worst",
    "concave_points_worst", "symmetry_worst", "fractal_dimension_worst"
]

print(len(feature_names))
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import pandas as pd

# Convert X_train into a DataFrame with feature names
X_train_df = pd.DataFrame(
    X_train,
    columns=feature_names
)

# Calculate VIF for every feature
vif_df = pd.DataFrame()

vif_df["Feature"] = X_train_df.columns
vif_df["VIF"] = [
    variance_inflation_factor(X_train_df.values, i)
    for i in range(X_train_df.shape[1])
]

# Sort from highest to lowest VIF
vif_df = vif_df.sort_values("VIF", ascending=False)

print("Variance Inflation Factor (VIF)")
print("--------------------------------")
print(vif_df.to_string(index=False))
```

```python
# ==========================================
# REDUCE FEATURES USING CORRELATION
# ==========================================

import numpy as np
import pandas as pd

# Convert training data to DataFrame
X_train_df = pd.DataFrame(X_train, columns=feature_names)
X_test_df = pd.DataFrame(X_test, columns=feature_names)

# Correlation matrix
corr_matrix = X_train_df.corr().abs()

# Take only upper triangle
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Find features having correlation > 0.95 with another feature
to_drop = [
    column
    for column in upper.columns
    if any(upper[column] > 0.95)
]

print("Number of features before:", X_train_df.shape[1])
print("Features to remove:", len(to_drop))
print("\nRemoved features:")
print(to_drop)

# Create reduced datasets
X_train_reduced = X_train_df.drop(columns=to_drop)
X_test_reduced = X_test_df.drop(columns=to_drop)

print("\nNumber of features after:", X_train_reduced.shape[1])
print("\nRemaining features:")
print(X_train_reduced.columns.tolist())
```

```python
# ==========================================
# EVALUATION
# ==========================================

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    roc_auc_score,
    brier_score_loss,
    log_loss
)

cm_reduced = confusion_matrix(
    y_test,
    y_pred_lr_reduced
)

tn_r, fp_r, fn_r, tp_r = cm_reduced.ravel()

accuracy_reduced = accuracy_score(
    y_test,
    y_pred_lr_reduced
)

sensitivity_reduced = tp_r / (tp_r + fn_r)

specificity_reduced = tn_r / (tn_r + fp_r)

brier_reduced = brier_score_loss(
    y_test,
    y_prob_lr_reduced[:, 1]
)

auc_reduced = roc_auc_score(
    y_test,
    y_prob_lr_reduced[:, 1]
)

logloss_reduced = log_loss(
    y_test,
    y_prob_lr_reduced
)

print("Reduced Logistic Regression Evaluation")
print("---------------------------------------")

print("\nConfusion Matrix:")
print(cm_reduced)

print("\nTrue Negatives :", tn_r)
print("False Positives:", fp_r)
print("False Negatives:", fn_r)
print("True Positives :", tp_r)

print("\nAccuracy       :", accuracy_reduced)
print("Sensitivity    :", sensitivity_reduced)
print("Specificity    :", specificity_reduced)
print("Brier Score    :", brier_reduced)
print("ROC-AUC        :", auc_reduced)
print("Log Loss       :", logloss_reduced)
```

## 13. Final Threshold Selection Using 5-Fold Stratified Cross-Validation

The final classification threshold is selected using **training data only**. Out-of-fold probabilities are generated with 5-fold stratified cross-validation, and Youden's J statistic is computed as `Sensitivity + Specificity - 1`. The threshold with the maximum Youden's J is selected and then locked before evaluating the untouched test set.

```python
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.base import clone
from sklearn.metrics import confusion_matrix

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Out-of-fold probabilities for TRAINING data only.
lr_cv = clone(log_reg)
y_train_prob_cv = cross_val_predict(
    lr_cv,
    X_train,
    y_train,
    cv=cv,
    method="predict_proba"
)[:, 1]

thresholds = [0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50]
cv_threshold_results = []

for threshold in thresholds:
    y_pred = (y_train_prob_cv >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_train, y_pred).ravel()

    sensitivity_cv = tp / (tp + fn)
    specificity_cv = tn / (tn + fp)
    accuracy_cv = (tp + tn) / len(y_train)
    youden_j = sensitivity_cv + specificity_cv - 1

    cv_threshold_results.append({
        "Threshold": threshold,
        "Accuracy": accuracy_cv,
        "Sensitivity": sensitivity_cv,
        "Specificity": specificity_cv,
        "Youden_J": youden_j,
        "FP": fp,
        "FN": fn
    })

cv_threshold_results = pd.DataFrame(cv_threshold_results)

best_row = cv_threshold_results.loc[cv_threshold_results["Youden_J"].idxmax()]
best_threshold = float(best_row["Threshold"])

print("5-Fold Cross-Validation Threshold Selection")
print("=" * 55)
display(cv_threshold_results.round(4))
print(f"Selected threshold: {best_threshold:.2f}")
print(f"Maximum Youden's J: {best_row['Youden_J']:.4f}")
```

### Sensitivity-Specificity Trade-off

```python
plot_df = cv_threshold_results.copy()

plt.figure(figsize=(8, 5))
plt.plot(plot_df["Threshold"], plot_df["Sensitivity"], marker="o", label="Sensitivity")
plt.plot(plot_df["Threshold"], plot_df["Specificity"], marker="o", label="Specificity")
plt.axvline(best_threshold, linestyle="--", label=f"Selected threshold = {best_threshold:.2f}")
plt.xlabel("Classification Threshold")
plt.ylabel("Score")
plt.title("Sensitivity-Specificity Trade-off (Training CV)")
plt.xticks(plot_df["Threshold"])
plt.ylim(0.85, 1.02)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 14. Final Evaluation on the Untouched Test Set

```python
from sklearn.metrics import precision_score, f1_score

# Test probabilities are generated only after the threshold has been locked.
y_test_prob_final = log_reg.predict_proba(X_test)[:, 1]
y_test_pred_final = (y_test_prob_final >= best_threshold).astype(int)

TN, FP, FN, TP = confusion_matrix(y_test, y_test_pred_final).ravel()

accuracy_final = (TP + TN) / (TP + TN + FP + FN)
sensitivity_final = TP / (TP + FN)
specificity_final = TN / (TN + FP)
precision_final = TP / (TP + FP)
npv_final = TN / (TN + FN)
f1_final = f1_score(y_test, y_test_pred_final)
fpr_final = FP / (FP + TN)
fnr_final = FN / (FN + TP)

print("FINAL TEST-SET EVALUATION")
print("=" * 55)
print(f"Locked threshold: {best_threshold:.2f}")
print(f"Test samples: {len(y_test)}")
print(f"TN = {TN}, FP = {FP}, FN = {FN}, TP = {TP}")
print(f"Accuracy: {accuracy_final:.4f}")
print(f"Sensitivity: {sensitivity_final:.4f}")
print(f"Specificity: {specificity_final:.4f}")
print(f"Precision: {precision_final:.4f}")
print(f"NPV: {npv_final:.4f}")
print(f"F1 Score: {f1_final:.4f}")
```

```python
from sklearn.metrics import brier_score_loss, log_loss, roc_auc_score

brier_final = brier_score_loss(y_test, y_test_prob_final)
logloss_final = log_loss(y_test, y_test_prob_final)
auc_final = roc_auc_score(y_test, y_test_prob_final)

print("PROBABILISTIC TEST-SET METRICS")
print("=" * 55)
print(f"Brier Score: {brier_final:.4f}")
print(f"Log Loss: {logloss_final:.4f}")
print(f"ROC-AUC: {auc_final:.4f}")
```
